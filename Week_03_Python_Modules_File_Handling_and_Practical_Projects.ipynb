```python
# Assignment 2

# MATH QUIZ

import math
import random

# Initialize count for scoring
count = 0

def add(num1, num2):
  question = f"What is the sum of {num1} and {num2}?"
  answer = int(input(question))
  correct_answer = num1 + num2
  return (question, answer, correct_answer)

def subtract(num1, num2):
  question = f"What is the difference between {num1} and {num2}?"
  answer = int(input(question))
  correct_answer = num1 - num2
  return (question, answer, correct_answer)

def multiply(num1, num2):
  question = f"What is the product of {num1} and {num2}?"
  answer = int(input(question))
  correct_answer = num1 * num2
  return (question, answer, correct_answer)

def divide(num1, num2):
  question = f"What is the quotient of {num1} and {num2}?"

  # Ensure num2 is not zero for division
  while num2 == 0:
    num2 = random.randint(1, 500)
  answer = input(question)
  correct_answer = f"{(num1 / num2):.2f}"
  return (question, answer, correct_answer)

def generate_random_question_and_answer():
  # Generate new random numbers for each question
  num1 = random.randint(1, 500)
  num2 = random.randint(1, 500)

  question_type = random.choice(['addition', 'subtraction', 'multiplication', 'division'])

  if question_type == 'addition':
    return add(num1, num2)
  elif question_type == 'subtraction':
    return subtract(num1, num2)
  elif question_type == 'multiplication':
    return multiply(num1, num2)
  elif question_type == 'division':
    return divide(num1, num2)

print("--- Starting the 5-Question Quiz ---\n")

# Loop 5 times to generate 5 questions
for i in range(5):
  print(f"Question {i+1}:")
  question, user_answer, correct_answer = generate_random_question_and_answer()

  if user_answer == correct_answer:
    print(f"Correct!! The answer is {correct_answer} Bravo")
    count += 1
  else:
    print(f"Wrong answer. The correct answer is {correct_answer}")
  print("\n")

print(f"Quiz finished! You got {count} out of 5 questions correct.")
```

```python

# Assignment 4

# DATE CALCULATOR

from datetime import datetime

birthday = input("when is your birthday ? ")

#user_birthdate = None

formats = [
    "%d %B",
    "%d, %B",
    "%d, %b",
    "%d/%m",
    "%d-%m-%Y",
    "%d %B %Y",
    "%m/%d/%Y",
    "%B %d, %Y",
    "%d/%m/%Y",
    "%Y-%m-%d"
]

# Add the current year to format, if year is not specified, since the actual birthyear is of little importance to the calculation

for format in formats:
    try:
        if "%Y" not in format:
            user_birthdate = datetime.strptime(
                birthday + f" {datetime.now().year}",
                format + "%Y")
        else:
             user_birthdate = datetime.strptime(birthday, format)

        break

    except ValueError:
        pass
    except Exception as e:
        print(f"An error occurred during this process: {e}")

if user_birthdate is None:
    raise ValueError("invalid date format")

present_date = datetime.now().date()
current_month = datetime.now().month
birth_month = user_birthdate.month

current_day = datetime.now().day
birth_day = user_birthdate.day

if (birth_month, birth_day) >= (current_month, current_day):
    user_birth_date = user_birthdate.replace(year=datetime.now().year).date()
else:
    user_birth_date = user_birthdate.replace(year=datetime.now().year + 1).date()

next_birthday = user_birth_date - present_date

print(f"Your next birthday is {next_birthday.days} days from now")
```

  # The difference between modules and packages

Module is simply a single file, that contains python codes which consists of functions, classes and variables, and can be imported into another python file. For example, math, random, etc are modules.

While packages on the other hand is a simple directory that has a collection of modules, that are grouped together. Traditionally, it contains an __init__.py file, but since Python 3.3, packages can exist without one; these are called namespace packages.  Modules  inside a package can only be accessed through that package. Simply put, it's a folder containing related modules

# ASSESSMENT 3

The math.pi constant gives the accurate value of pi(π). The purpose of this constant is to aid mathematical calculations, such as circumference of a circle, area, volume, etc.

  It simply make these mathematics calculations easier, and more accurate, especially due to the fact that pi is a floating point value.

```python
# Using the math.pie constant

# to get area of a circle

import math

def area(radius):
   x = math.pi * (radius ** 2)
   print(f"Area = {x:0.2f}")

area(5)

# to get circumference of a circle

def circumference(radius):
  a = 2 * math.pi * radius
  print(f"circumference = {a:0.2f}")

circumference (7)

# To get volume of a sphere

def volume(radius):
   b = (4 / 3) * math.pi * (radius ** 3)
   print(f"volume = {b:0.3f}")
volume(9)
```

# ASSESSMENT 8

The difference between "import module", and "from module import function" is that;

Import module imports the entire module or python file, with all the functions and objects in it, which are accessed by using the module name is prefix. e.g: math.sqrt. It keeps things organized, and avoid naming conflicts. while;

From module import function, is specific. It goes into the module, looks for that particular function, or attribute listed, and imports it. It calls the function directly, having high risk of collision. It's more convenient when you only need one or two things from a module.

# ASSESSMENT 10

Virtual environment is an isolated python environmenttthat allows you to manage dependencies (external code packages, or library that the project requires to run, build or develop properly) for  different  projects independently used to avoid conflicts and clash between project dependencies.

  In simple terms, it's a separate container for each python projects, having its own interpreter, packages, library, etc.

```python
# ASSESSMENT 11
import datetime

def next_event():
   today = input("what is the current date? ")
   event = input("when is your next event / deadline? ")

   formats = [
          "%d,%B,%Y",
          "%d/%B/%Y",
          "%d-%m-%Y",
          "%d/%m/%Y",
          "%m/%d/%Y",
          "%Y/%m/%d"
            ]

   current_date = None
   event_date = None

   for format in formats:
     try:
       current_date = datetime.datetime.strptime(today, format)
       event_date = datetime.datetime.strptime(event, format)
       break

     except ValueError:
       continue

     except:
       print("An error occurred")

   if current_date is None or event_date is None:
     raise ValueError("invalid date format")


   difference  = event_date - current_date

   if current_date > event_date:
      print(f"You are {abs(difference.days)} days behind schedule. You need to buckle up.")
   else:
      print(f"You have {difference.days} days left.")

next_event()
```
