````python
# SHOPPING CART

supermarket = {"Keyboard":350, "mouse":450,
               "Bag":20, "Cornflake":5,
               "Chin_chin":4, "Apple":1.5,
               "Milk":5, "Cloth": 25,
               "Laptop":4500, "Orange":2,"Rice":25,
               "Phone":250, "Juice":5, "Cap" :25,
               "Shawarma":80, "Pizza":90, "Yam":75) }


items_input = input("Enter the products you want to buy: ")

desired_items = [item.strip().title() for item in items_input.split(',')]


actual_shopping_cart = {}
total_price = 0.0

for item_name in desired_items:

    if item_name in supermarket:
        price = supermarket[item_name]
        actual_shopping_cart[item_name] = price
        total_price += price
    else:
        # when items is not found
        print(f"'{item_name}' not found in the supermarket.")

print("\nMy Shopping Cart: ")
if actual_shopping_cart:
  for item, price in actual_shopping_cart.items():
      print(f"{item}: ${price}")
else:
    print("No valid items were added to the cart.")

print(f"Total to pay: ${total_price}")
````

````python
# GUESSING GAME

import random
number = random.randint(1, 100)
guess = 0

while guess != number:
  guess = int(input("Guess the next number: "))
  if guess > number:
    print("Wrong choice. Enter a lower number")
  elif guess < number:
    print("Wrong choice, Enter a higher number")
  elif guess < 0:
    print("wrong input")
  else:
    print("Congratulations!!! You did it")
````

# THE DIFFERENCE BETWEEN LIST AND TUPLE

 List and tuples are used to represent series of comma seperated items and values, but they have lots of difference and use cases, which are:
 1. **List** are mutable (you can edit it after creation), but **tuples** are immutable and unchangeable.
 2. List values are stored within square brackets, while tuples are stored within parentheses.
 3. List are more prone to unexpected errors than tuples, this is due to its mutability.
 4. List comsumes more memory, and has several built-in methods, such as append, del,insert, etc. Tuples uses lesser memory compared to list, and has lesser built-in method because of immutability.
 5. List cannot be used as dictionary keys, but tuples can be used.
 6. Tuples uses heterogeneous Sequence of elements accessed via unpacking and indexing. List uses homogeneous sequence and  are accessed by iterating over the list.
 7. Iteration in list is time consuming, but faster in tuple.
 An example of a list is: x = [10, 20, 30]
 An example of a tuple is: y = (1, 2, 3, 4)

# DIFFERENCE BETWEEN LOCAL AND GLOBAL VARIABLES

Local and global variables, are the 2 types of variable in n python. The difference between both are;

LOCAL VARIABLE: They are defined within a function and can only be accessible inside that block or function. Memory for these variable is allocated on stack, and each function call has its own copy of variable. It is created when the function is called, and destroyed when the function returns.

WHILE;

GLOBAL VARIABLE: They are defined outside the block or function, and can be accessed throughout the program. Memory are allocated on data segment, and all function shares the same copy of variable. It is created when the program starts and destroyed when the program ends.

# DIFFERENCE BETWEEN FOR AND WHILE LOOP WITH EXAMPLES

for loop: it is used when you know how many times you want to repeat something, or a particular action. For example going to the supermarket with of list of the exact things you want to buy

While loop: it is used when we don't know how many times we will repeat something. It depends on a condition that may change during execution. For example, you don't know have far you can run. You can only stop when tired.

# CODE EXAMPLES OF FOR AND WHILE LOOP

````python
# for loop
items = ["Beverages", "Rice", "Toothpaste"]

Count = 1
for item in items:
  print(Count, "Buying", item)
  Count += 1
````

````python
# while loop

i = 1
while i < 7:
  print(i)
  i += 1
````

# PYTHON EXCEPTIONS AND CODE THAT HANDLES DIVISION BY ZERO

Exception is a event that occurs during the execution of a program, that distrusts the normal flow of instruction. It prevents the program from crashing.

Exceptions can be handled using 'try', 'except', & optionally 'finally' blocks.

If an exception occurs in the try block, the code in exception is executed.


# CODE THAT HANDLES DIVISION BY ZERO

````python
try:
  numerator = 15
  denominator = 0
  result = numerator / denominator

except ZeroDivisionError:
  print("Error: can't divide by Zero!")
else:
  print("Result: ", result)
finally:
  print("try another value")
````

# A PROGRAM THAT CHECKS IG A STRING IS A PALINDROME

Palindrome is something that reads the same forward and backward

````python
# To check if a string is a palindrome;

string = input("Enter a string: ")
string = string.lower()

if (string == string[::-1]):
    print("This string is a palindrome")
else:
    print("This strong is not a palindrome")
````
