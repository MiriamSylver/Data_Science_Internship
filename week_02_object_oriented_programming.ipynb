- Create a base class `BankAccount` with `owner`, `balance`, and methods `deposit()` and `withdraw()`.
- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.
- Demonstrate deposits, withdrawals, and interest application.

````python
class BankAccount:
  def __init__(self, owner, balance):
    self.owner = owner
    self.balance = balance

# Adding Money to the account(Deposit)
  def deposit(self, add_amount):
    self.balance += add_amount
    return self.balance

# Removing money from the account(withdrawal)
  def withdraw(self, remove_amount):
    if self.balance >= remove_amount:
      self.balance -= remove_amount
      return self.balance
    else:
      return ("insufficient funds!!")

# Interest gotten from savings account
class SavingsAccount(BankAccount):
  def __init__(self, owner, balance, interest_rate, time):
    super().__init__(owner, balance)
    self.balance = balance
    self.interest_rate = interest_rate
    self.time = time

  def add_interest(self):
    self.balance += (self.balance * self.time * (self.interest_rate / 100))
    return self.balance


# Code implementation
MyAccount = BankAccount("John Harrison", 20)
print(f"Greetings {MyAccount.owner}, Your account balance is: ${MyAccount.balance}")

deposit_amount = MyAccount.deposit(50)
print(f"You just deposited ${deposit_amount}. \nCurrent balance: ${MyAccount.balance}")

withdrawal_amount = 25
MyAccount.withdraw(withdrawal_amount)

if withdrawal_amount <= MyAccount.balance:
  print(f"You have removed ${withdrawal_amount} from your account. \n Current balance: ${MyAccount.balance}")
else:
  print (f"You have insufficient funds. Your Current balance is ${MyAccount.balance}. Please add more money")


# Creating a savings account
My_savings_account = SavingsAccount("John Harrison", 100, 29, 5)
print(f"Greetings {My_savings_account.owner}. You just opened a savings account at an interest rate of {My_savings_account.interest_rate}% per annum.\nBalance: ${My_savings_account.balance}")

# Adding interest to savings account
Current_value = My_savings_account.add_interest()
print(f"Your current balance is ${Current_value} after {My_savings_account.time} years interest")
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
  def __init__(self, n):
    self.n = n

  def __iter__(self):
    return self

  def __next__(self):
    if self.n >= 1:
      x = self.n
      self.n -= 1
      return x
    else:
      raise StopIteration

count_down = CountDown(10)

for x in count_down:
  print(x)
````

# METHOD RESOLUTION ORDER

Method Resolution Order (MRO) defines the order in which Python searches for a method in a class and its parent classes. Python uses a C3 linearization to compute MRO. It checks the child class first, and if the class inherits multiple, it searches them from left to right, that means the parent at the left will be implemented first.

For Example, in the task above, it checked the duck class, but no value was given, so it moved on to check the flyer method, which executed  first,  because  its at the left.

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
# FIBONACCI SEQUENCE

class Fibonacci:
    def __init__(self, limit):
        self.limit = limit
        self.count = 0

        self.prev_1 = 0
        self.prev_2 = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.limit:
            raise StopIteration

        if self.count == 0:
            self.count += 1
            return self.prev_1
        elif self.count == 1:
            self.count += 1
            return self.prev_2


        current = self.prev_1 + self.prev_2


        self.prev_1 = self.prev_2
        self.prev_2 = current

        self.count += 1
        return current

# code implementation
fib_sequence = Fibonacci(20)
for number in fib_sequence:
    print(number)
````

````python
# MY LIBRARY MANAGEMENT SYSTEM

# Creating a Library

# Who can add book to a library
class User:

  def __init__(self, name, is_admin=False):
    self.name = name
    self.is_admin = is_admin

# The Library

class Library:
  def __init__(self):
    self.books = []
    self.__library_name = "sylver's_library"

  def get_description(self):
    print(f"Welcome to {self.__library_name}. This is a very large academic database where members can access all kinds of book, grow they knowledge and make impact")

  def access_library_name(self):
    print (self.__library_name)

# Permission to add book

  def add_book(self, user, book):
    if not user.is_admin:
      raise PermissionError(f"Access denied! {user.name} cannot add books.")

    self.books.append(book)
    return (f"Book: '{book.title}' added by admin")



# Iterating through the library
  def __iter__(self):
    self._current_book_index = 0
    return self

  def __next__(self):
    if self._current_book_index < len(self.books):
        book = self.books[self._current_book_index]
        self._current_book_index += 1
        return book
    else:
        raise StopIteration

# Search by title or author
  def search(self, keyword):
    found_books = []
    for book in self.books:
      if keyword.lower() in book.title.lower() or keyword.lower() in book.author.lower():
        found_books.append(book)
    return found_books

 # Book Management session

class Book:
  def __init__(self, title, author, year):
    self.title = title
    self.author = author
    self.year = year
    self.is_borrowed = False

  def __str__(self):
    status = "Borrowed" if self.is_borrowed else "Available"
    return f"[{self.title} by {self.author} — Status: {status}]"

  def get_description(self):
    return(f"{self.title} written by {self.author}, and published in {self.year}")


class EBook(Book):
  def __init__(self, title, author, year, file_size, format):
    super().__init__(title, author, year)
    self.file_size = file_size
    self.format = format


  def get_description(self):
      return(f"""{self.title}
Written by {self.author}
Published in {self.year}
File size: {self.file_size}
Format: {self.format}""")

class PrintedBook(Book):
  def __init__(self, title, author, year, weight, dimension):
    super().__init__(title, author, year)
    self.weight = weight
    self.dimension = dimension

  def get_description(self):
      return(f"""{self.title}, written by {self.author}, published in {self.year}
Weight: {self.weight}
Dimension: {self.dimension}""")


# Members Management session

class Member:
  _total_borrowed_books_count = 0

  def __init__(self, name, member_id):
    self.name = name
    self.member_id = member_id
    self._borrowed_books = []

  def borrow_a_book(self, book):
    if not isinstance(book, Book):
      raise ValueError("This book, is currently not available.")

    if not book.is_borrowed:
        book.is_borrowed = True
        self._borrowed_books.append(book)
        Member._total_borrowed_books_count += 1
        print(f"'{book.title}' borrowed by {self.name}. Total borrowed: {Member._total_borrowed_books_count}")
    else:
        print(f"'{book.title}' is already borrowed.")

  def return_a_book(self, book):
      if book in self._borrowed_books:
          book.is_borrowed = False
          self._borrowed_books.remove(book)
          Member._total_borrowed_books_count -= 1
          print(f"'{book.title}' returned by {self.name}. Total borrowed: {Member._total_borrowed_books_count}")
      else:
          print(f"'{book.title}' was not borrowed by {self.name}.")

  def get_borrowed_books(self):
      return self._borrowed_books

  def __iter__(self):
     self._borrowed_books_iter_index = 0
     return self

  def __next__(self):
    if self._borrowed_books_iter_index < len(self._borrowed_books):
      book = self._borrowed_books[self._borrowed_books_iter_index]
      self._borrowed_books_iter_index += 1
      return book

    else:
      raise StopIteration

  # General borrow_book method
  def borrow_book(self, book_title):
    return f"You can always borrow book from this library. Book: {book_title}"


class StudentMember(Member):
  def __init__(self, name, member_id, grade_level):
    super().__init__(name, member_id)
    self.grade_level = grade_level
    self._borrowing_limit = 5

# Borrowing limit
  def borrow_book(self, book):
    if not isinstance(book, Book):
      raise ValueError("Sorry, this book is currently unavailable.")

    if len(self._borrowed_books) >= self._borrowing_limit:
      return(f"Sorry dear student {self.name}, you have reached your borrowing limit of {self._borrowing_limit} books.")
    else:
      self.borrow_a_book(book)
      return (f"{self.name} (Student) successfully borrowed '{book.title}'.")

class TeacherMember(Member):
  def __init__(self, name, member_id, department):
    super().__init__(name, member_id)
    self.department = department
    self._borrowing_limit = 10

# Borrowing limit
  def borrow_book(self, book):
    if not isinstance(book, Book):
      raise ValueError("Sorry, this book is currently unavailable.")

    if len(self._borrowed_books) >= self._borrowing_limit:
      return(f"Sorry dear Teacher {self.name}, you have reached your borrowing limit of {self._borrowing_limit} books.")
    else:
      self.borrow_a_book(book)
      return f" {self.name} (Teacher) successfully borrowed '{book.title}'."


# code implementation

# Create book

book1 = Book("Things Fall Apart", "Chinua Achebe", 1958)

book2 = Book("The Hobbit", "J.R.R. Tolkien", 1937)

book3 = Book("Purple Hibiscus", "Chimamanda Ngozi Adichie", 2003)

book4 = Book("Half of a Yellow Sun", "Chimamanda Ngozi Adichie", 2006)

book5 = Book("Atomic Habits", "James Clear", 2018)


# Demonstrate Book printing
print(book1)
print(book1.get_description())

# Demonstrate EBook and PrintedBook
ebook1 = EBook("Digital Fortress", "Dan Brown", 1998, "2.5MB", "PDF")
printedbook1 = PrintedBook("The Lord of the Rings", "J.R.R. Tolkien", 1954, "1.2kg", "6x9 inches")
print(ebook1.get_description())
print(printedbook1.get_description())

# Create a library instance
my_library = Library()

# Create users (admins for adding books)
admin_user = User("Admin John", is_admin=True)
regular_user = User("Regular Jane")

# Add books to library
try:
  print(my_library.add_book(admin_user, book1))
  print(my_library.add_book(admin_user, ebook1))
  print(my_library.add_book(admin_user, printedbook1))
except PermissionError as e:
  print(e)

# Iterate over books in the library
print("\nBooks in Library:")
for book in my_library:
  print(book)

# Search for books
print("\nSearching for 'Tolkien':")
found_by_tolkien = my_library.search("Tolkien")
for book in found_by_tolkien:
  print(book)

# Create members
student1 = StudentMember("Alice", "S001", "Grade 10")
teacher1 = TeacherMember("Mr. Smith", "T001", "Science")

# Members borrow books
print("\nMember borrowing actions:")
print(student1.borrow_book(book1))
print(student1.borrow_book(book2))
print(student1.borrow_book(book3))
print(student1.borrow_book(book4))
print(student1.borrow_book(book5))
print(student1.borrow_book(ebook1)) # Student tries to borrow beyond limit

print(teacher1.borrow_book(book1)) # Teacher tries to borrow an already borrowed book
print(teacher1.borrow_book(ebook1))

# Iterate over a member's borrowed books
print(f"\n{student1.name}'s borrowed books:")
for book in student1:
  print(book)

print(f"\n{teacher1.name}'s borrowed books:")
for book in teacher1:
  print(book)

# Demonstrate total borrowed count
print(f"\nTotal books borrowed by all members: {Member._total_borrowed_books_count}")

# Member returns a book
student1.return_a_book(book1)
print(f"Total books borrowed by all members after return: {Member._total_borrowed_books_count}")

# Access library name (getter)
print("\nLibrary name:")
my_library.access_library_name()
````
